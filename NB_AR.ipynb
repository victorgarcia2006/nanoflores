# 1. Convertir archivos .eps a .png

```python
import os
from PIL import Image

init_dir= 'D:\OneDrive - Instituto Politecnico Nacional\Investigacion\Particulas_magneticas\V2025-2'
# os.chdir(init_dir)

print( os.getcwd() )

# Select experiment 
exp = 'Aspect_Ratio'


os.chdir(exp+'\eps')
print( os.getcwd() )

#current_dir = os.getcwd()
#print(current_dir)
#os.chdir(r'eps')


for file in os.listdir():
    if file.endswith('.eps'):
        #print(file, os.path.splitext(file) )
        print(f"Convirtiendo {file} \t en \t {os.path.splitext(file)[0]}.png")
        with Image.open(file) as img:
            img = img.convert('RGB')
            o_file = os.path.splitext(file)[0] + ".png"
            img.save(o_file,"PNG")

src_dir= os.getcwd()
dest_dir= f"D:\OneDrive - Instituto Politecnico Nacional\Investigacion\Particulas_magneticas\V2025-2\{exp}\png"
for file in os.listdir():
    if file.endswith('.png'):
        src_path = os.path.join(src_dir, file)
        dst_path = os.path.join(dest_dir, file)
        os.rename(src_path, dst_path)
        print(f" Moviendo {file} ")

os.chdir(init_dir)

print("codigo finalizado con éxito")
```

# 2. Generar curvas de histéresis junto al esquema geométrico
